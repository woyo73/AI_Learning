```python
# 딕셔너리를 선언합니다.
dictionary = {}

# 요소 추가 전에 내용을 출력해 봅니다.
print("요소 추가 이전:", dictionary)

dictionary['name'] = '새로운 이름'
dictionary['head'] = "새로운 정신"
dictionary['body'] = "새로운 몸"

# 딕셔너리에 요소를 추가합니다.
if 'name' in dictionary:
  print("aa")

# 출력합니다.
print("요소 추가 이후 : ", dictionary)
```

```python
# 딕셔너리 요소 제거하기

# 딕셔너리를 선언합니다.
dictionary =  {
  'name': '새로운 이름', 
  'head': '새로운 정신', 
  'body': '새로운 몸'
  }

# 요소 제거 전에 내용을 출력해봅니다.
print("제거전 :", dictionary)

# 딕셔너리의 요소를 제거합니다.
del dictionary['name']

# 요소 제거 후에 내용을 출력해봅니다.
print("제거후 :", dictionary)
```

```python
# 키가 존재하는지 확인하고 값에 접근하기

# 딕셔너리를 선언합니다.
dictionary = {
  'name': '새로운 이름', 
  'head': '새로운 정신', 
  'body': '새로운 몸'
}

# 사용자로부터 입력을 받습니다.
key = input("> 접근하고자 하는 키: ")

# 키값 존재 여부 확인 
if key in dictionary:
  print(dictionary[key])
else:
  print("키가없어요!!")

for key, value in dictionary.items():
  print(f"{key} {value}")
```

```python
# 키가 존재하지 않을 때 None을 출력하는지 확인하기

# 딕셔너리를 선언합니다.
dictionary = {
  'name': '새로운 이름', 
  'head': '새로운 정신', 
  'body': '새로운 몸'
}

# 존재하지 않는 키에 접근해 봅니다.
value = dictionary.get("type")
print(value)

if value == None:
  print("없는 키!!!")



# None 확인 방법
```

```python
# for 반복문과 딕셔너리

# 딕셔너리를 선언합니다.
dictionary = {
  'name': '새로운 이름', 
  'head': '새로운 정신', 
  'body': '새로운 몸'
}


# for 반복문을 사용합니다
# 키값만 얻어오기 
for key in dictionary:
  print(f"{key} : {dictionary[key]}")

# 키 벨류 같이 가져오기 
for key, value in dictionary.items():
  print(f"{key} : {value}")
```
