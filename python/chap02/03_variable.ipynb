## 3. 변수와 입력

 * 변수 : 데이터를 저장하기위 해 할당된 메모리 저장소 공간에 붙인 이름

```python
data1 = 3
data2 = 5
print(data1+data2)
```

```python
data1 = 7
print(data1+data2)
```

```python
# 메모리 할당 요청 방법과 데이터 저장 방법 
pi = 3.141592
r = 10

# 메모리에 접근(참조) 방법
print("원주율 =", pi)
print("반지름 =", r)

print("원의 넓이 = ", pi * r * r)
print("원의 둘레 = ", 2 * pi * r)
```

```python
# 입력 합수 input()
pi = 3.141592

while True:
  try:
    r = float(input("반지름의 값을 입력하세요>"))  
  except ValueError as e:
    print("숫자나 실수만 입력해라!!")  
    print(e)  
  else:
    print("원의 넓이 = ", pi * r * r)
    print("원의 둘레 = ", 2 * pi * r)
    break
```

```python
try:
  age = int(input("나이를 입력하세요>"))
except ValueError:
  print("숫자를 입력하세요!")
else:
  print(f"당신의 나이는 {age} 입니다.")


```
