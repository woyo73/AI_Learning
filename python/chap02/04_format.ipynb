## 4. 숫자와 문자열의 다양한 기능
 ### - 문자열의 format() 함수

```python
print("이름 {1}  나이 {0}".format("우영", 11))
```

```python
"{} {}".format(10, 20)
```

```python
won = input("가격을 입력하세요>")

if (not won) == False:
  print("{}만원".format(won))
```

```python
print("{:+05d}".format(-52))
print("{:5d}".format(52))
print("{:10d}".format(52))
print("{:010d}".format(-52))
print("{: d}".format(52))
print("{: d}".format(-52))
print("{:=+05d}".format(52))
print("{:=+05d}".format(-52))
print("{:=+0.2f}".format(52))
print("{:=+0.2f}".format(-52))

print("{:g}".format(52.1253))


print("abcd")
print("abcd".upper())
print("abcd".lower())


print("  abcd  ".strip())
print("   abcd  ".lstrip())
print("   abcd  ".rstrip())


```
