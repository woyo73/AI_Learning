# 자료형 ( Data Type )

```python
print("안녕하세요")
print(
  """\
  dddd
  dddd\
  """)
print(type("안녕하세요"))
print(type(846))
print(type(1.12))
print(type(True))

print("철수가 \"안녕\"이라고 인사를 했어.")
```

### - 문자열 연산자

```python
# 문자열 연산자 : +
print("안녕" + "하세요.")
```

```python
print(3*5)
print("안녕하세요" * 3)
```

```python
# 문자 선택 연산자 ( 인덱싱 ) : []
print("안녕하세요"[-1])


# 문자열 범위 선택 연산자 ( 슬라이싱 ) : []
print("안녕하세요"[0:1])
print("안녕하세요"[0:2])
print("안녕하세요"[0:5:2])

# 문자열 길이 
print(len("안녕하세요"))

```
