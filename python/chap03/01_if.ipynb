### * 조건문의 기본 사용


```python
# 입력을 받습니다.
input_value = int(input("정수 입력>"))

if input_value > 0:
  print("양수입니다.")  

if input_value < 0:
  print("음수입니다.")

if input_value == 0:
  print("0 입니다.")

print("사용자가 입력한 값: ", input_value)
print("프로그램 종료")
```

### 날짜/시간 활용하기

```python
# 날짜/시간과 관련된 기능을 가져옵니다. 
import datetime

# 현재 날짜/시간을 구합니다. 
now = datetime.datetime.now()

# 출력 
print("{}년 {}월 {}일 {}시 {}분 {}초".format(
  now.year, 
  now.month, 
  now.day,
  now.hour, 
  now.minute,
  now.second))

```

```python
import datetime

now = datetime.datetime.now()

# 봄 3~5
if 3 <= now.month <= 5:
  print("봄")
# 여름 6~8
elif now.month <= 8:
  print("여름")
# 가을 9~11
elif now.month <= 11:
  print("가을")
# 겨울 12~2
else:
  print("겨울")
```
