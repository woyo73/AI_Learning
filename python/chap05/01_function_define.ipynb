```python
# 기본 매개변수
def printPlayer(name, hp=100, mp=50, lv = 1):
  print(f"name:{name} hp:{hp} mp:{mp} lv:{lv}")

# 함수를 호출합니다.
printPlayer("우영")
printPlayer("우영", 50)

# 기본 매개변수(default argument) 뒤에는 일반 매개변수(Non-default argument)가 올 수 없다.

def attack(name, *damages, bonus = 0):
  total = sum(damages) + bonus
  print(f"{name} 총 데미지: {total}")

attack("우영", 10, 20, 30, 1, bonus=2000)



```

```python
# 키워드 매개변수
def print_player(name, hp, mp):
  print(name, hp, mp) 

# 함수를 호출합니다.
print_player(mp=100, hp=1000, name = "우영")

print_player("우영", mp=999, hp=222)


```

```python
# 기본 매개변수 중에서 필요한 값만 입력하기
# 여러 함수 호출 형태
def print_player(name="전사", hp = 100, mp = 50):
  print(name, hp, mp)

# 1) 기본 형태
print_player("마법사", 80, 150)

# 2) 키워드 매개변수로 모든 매개변수를 지정한 형태
print_player(name="우영", hp=999, mp=111)


# 3) 키워드 매개변수로 모든 매개변수를 마구잡이로 지정한 형태
print_player(hp=999, name="우영", mp=111)


# 4) 키워드 매개변수로 일부 매개변수만 지정한 형태
print_player(mp=200)
```

```python
# 범위 내부의 정수를 모두 더하는 함수

# 함수를 선언합니다.
def add_range(start, end) :    
  return sum(range(start, end+1))

# lambda_add = lambda start, end: sum(range(start, end+1))

# 함수를 호출합니다.
print("0 to 100:", add_range(0, 100))
print("0 to 1000:", add_range(0, 1000))
print("50 to 100:", add_range(50, 100))
print("500 to 1000:", add_range(500, 1000))
```

```python
# 기본 매개변수와 키워드 매개변수를 활용해 범위의 정수를 더하는 함수

# 함수를 선언합니다.
# def sum_all(start = 0, end = 100, step = 1):
#   total = 0

#   for i in range(start, end + 1, step):
#     total += i

#   return total

sum_all = lambda start = 0, end = 100, step = 1: sum(range(start, end +1, step))

# 함수를 호출합니다.
print("A.", sum_all(0, 100, 10))
print("B.", sum_all(end = 100))
print("C.", sum_all(end = 100, step = 2))
```
