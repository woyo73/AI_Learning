```python
partys = [
      {"name": "Knight", "hp": 100, "attack": 25},
      {"name": "Archer", "hp": 0, "attack": 35},
      {"name": "Mage", "hp": 60, "attack": 50},
      {"name": "Healer", "hp": 80, "attack": 10}
  ]  

alive_names = [part["name"] for part in partys if part["hp"] > 0]
print(alive_names)

strong_names = [part["name"] for part in partys if part["attack"] >= 30]
print(strong_names)

status = ["생존" if part["hp"] > 0 else "사망" for part in partys]
print(status)
```

```python
names = ["knight", "Mage", "Archer", "Healer"]
attacks = [25, 35, 50, 10]

players = list(zip(names, attacks))

ranked = sorted(
  players, 
  key=lambda item:item[1],
  reverse=True
)

filtered = [player for player in ranked if player[1] >= 30]

for rank, (name, score) in enumerate(filtered, start=1):  
    print(f"{rank}위 : {name} ({score}점)")
```
