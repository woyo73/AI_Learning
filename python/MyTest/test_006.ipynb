```python
damages = [10, 20, 30, 40, 50]

# 1. generate_damage(damages) 제너레이터 함수 정의
def generate_damage(damages):
  for damage in damages:
    yield damage


# 2. next()로 첫 번째, 두 번째 공격 출력
gen = generate_damage(damages)
print(next(gen))
print(next(gen))


# 3. 새 제너레이터를 만들어 for 반복문으로 전체 출력
gen = generate_damage(damages)
for damage in gen:
  print(damage)

# 4. 제너레이터 표현식으로 피해량 두 배 출력
result = (x * 2 for x in damages)
print("result:", list(result))

```

```python
from contextlib import contextmanager

# 1. BattleSession 클래스 정의
class BattleSession:
  def __enter__(self):
    print("전투시작")    
    return self

  def __exit__(self, exc_type, exc, tb):
    print("전투종료")    

# 2. with BattleSession() 사용
with BattleSession() as battle:
  print("knight 공격!!")    


# 3. @contextmanager를 이용한 battle_context 정의
@contextmanager
def battle_context():
  print("전투시작")

  try:
    yield
  finally:
    print("전투종료")

# 4. with battle_context() 사용
with battle_context():
  print("mage 공격!!")    


```

```python
from dataclasses import dataclass

@dataclass
class Player:
  name: str
  hp: int
  attack: int

  def __post_init__(self) -> None:
    if self.hp < 0:
      raise ValueError("hp는 음수일 수 없습니다.")
    if self.attack < 0:
      raise ValueError("attack는 음수일 수 없습니다.")

player = Player("woo", 10, 2)
```

```python
from dataclasses import dataclass, field

@dataclass
class Player:
  name: str
  hp: int = 100
  attack: int = 100
  inventory: list[str] = field(default_factory=list)

  def take_damage(self, damage: int) -> None:
    self.hp = max(0, self.hp - damage)    

  def is_alive(self) -> bool:
    if self.hp > 0:
      return True
    else:
      return False

  def add_item(self, item: str) -> None:
    self.inventory.append(item)

  def __post_init__(self) -> None:
    if self.hp < 0:
      raise ValueError("hp는 음수일 수 없습니다.")


player = Player("Knight", 100, 1000)

player.take_damage(30)
player.add_item("Sword")

print(player.hp)
print(player.is_alive())
print(player.inventory)




```

```python



import asyncio 

async def download(name):
  print(f"{name} 시작")
  await asyncio.sleep(2)
  print(f"{name} 완료")

async def main():
  await asyncio.gather(
    download("파일 a"), 
    download("파일 b")
  )

await main()

print("전부 완료!!")
```

```python
import asyncio

monsters = ["Slime", "Goblin", "Dragon"]

async def fetch_monster(name: str) -> str:
  await asyncio.sleep(1)
  return f"{name} 데이터 로드 완료"

async def main():
  results = await asyncio.gather(
      *(fetch_monster(monster) for monster in monsters)
  )

  for result in results:
    print(result)

await main()


```
