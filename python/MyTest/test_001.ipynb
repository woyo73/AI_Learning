```python
class Player : 
  def __init__(self, name, hp, attack):
    self.name = name
    self.hp = hp
    self.attack = attack

  def take_damage(self, damage):
    self.hp = max(0, self.hp - damage)

  def show_info(self):
    print(f"{self.name} hp:{self.hp} attack:{self.attack}")  


class Warrior(Player) :
  def __init__(self, name, hp, attack):
    super().__init__(name, hp, attack)

  def power_attack(self):
    return self.attack * 2


warrior = Warrior("Knight", 100, 25)
warrior.take_damage(30)
warrior.show_info()

powerAttack = warrior.power_attack()

print(f"강력한 공격 피해량 : {powerAttack}")




  

```
