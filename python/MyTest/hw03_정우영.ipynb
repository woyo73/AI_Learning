* 문제 1 
  - 피타고라스의 정리 a^2 + b^2 = c^2 에서 사용자로 부터 밑변의 길이와, 높이의 값을 입력 받았을때, 빗변의 길이를 출력하는 프로그램을 작성해보세요.


 


```python
while True:
  try:  
    a = float(input("밑변의 길이를 입력하세요>"))    
  except ValueError:
    print("숫자만 입력해주세요")
    continue    

  try:
    b = float(input("높이의 값을 입력하세요>"))
  except ValueError:
    print("숫자만 입력해주세요")
    continue

  print(f"빗변의 길이는 {(a**2+b**2)**0.5}")
  break
```

* 문제 2
  - 정수 3개를 입력받아 (input() 함수 이용) 가장 큰 수를 출력하세요.


```python
numbers = []

while True:
  numbers.clear()

  try:
    numbers.append(int(input("첫번째 정수를 입력하세요>")))
  except ValueError:
    print("숫자만 입력해주세요!")
    continue

  try:
    numbers.append(int(input("두번째 정수를 입력하세요>")))
  except ValueError:
    print("숫자만 입력해주세요!")
    continue

  try:
    numbers.append(int(input("세번째 정수를 입력하세요>")))
  except ValueError:
    print("숫자만 입력해주세요!")
    continue

  result = sorted(numbers, reverse = True)
  print(result[0])
  break
```

* 문제 3
  - 연도를 입력 받아 윤년인지 판별하세요.
  - 윤년의 조건 ( 400으로 나누어 떨어지면 윤년, 또는 4로 나누어 떨어지고 100으로는 나누어 떠러지지 않으면 윤년)

```python
while True:
  try:
    year = int(input("연도를 입력 하세요>"))
  except ValueError:
    print("숫자만 입력해주세요!")
    continue

  if year % 400 == 0 or (year % 4 == 0 and year % 100 != 0):
    print("윤년")
  else:
    print("윤년 아님")

  break
```

* 문제 4
  - 사용자에게 태어난 연도의 값을 입력받아, 띠를 출력하는 프로그램을 작성하세요.

```python
animals = [ "쥐", "소", "호랑이", "토끼",
            "용", "뱀", "말", "양",
            "원숭이", "닭", "개", "돼지"]

while True:
  try:
    year = int(input("태어난 연도를 입력하세요>"))
  except ValueError:
    print("숫자만 입력해주세요!")
    continue

  index = (year - 4) % 12
  print(f"당신의 띠는 {animals[index]}띠 입니다.")
  break 

```
