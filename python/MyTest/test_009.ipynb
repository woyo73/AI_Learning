# 텍스트·바이너리 데이터와 모듈 관리

```python
text = "안녕하세요"

binary_data = text.encode("utf-8")

print(binary_data)
print(type(binary_data))

decoded_text = binary_data.decode("utf-8")

print(decoded_text)
print(type(decoded_text))
```

```python
data = "Python Binary Data!!우리나라".encode("utf-8")

with open("sample.bin", "wb") as file:
  file.write(data)

with open("sample.bin", "rb") as file:
  result = file.read().decode("utf-8")

print(result)
# with open()
```
