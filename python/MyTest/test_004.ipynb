```python
import importlib
import player
importlib.reload(player)

from player import Player 
import json

player = None
monster = None

while True:
  print("===== RPG 캐릭터 관리 =====")
  print("1. 캐릭터 생성")
  print("2. 캐릭터 정보")
  print("3. 몬스터 전투")
  print("4. 캐릭터 회복")
  print("5. 게임 저장")
  print("6. 게임 불러오기")
  print("0. 종료")
  print()
  print("메뉴를 선택하세요:")

  try:
    select_menu = int(input())
  except ValueError:
    print("숫자를 입력하세요!")
    continue

  if select_menu == 0:
    break

  elif select_menu == 1: 
     print("이름 : ")
     name = input().strip()
     if not name:
        print("캐릭터 이름을 일력하세요!")
     else:
      player = Player(name, 100, 10)
      print("캐리터 생성 완료!!")     

  elif select_menu == 2:
     if player is None:
        print("생성된 캐릭터가 없습니다.")
     else:
        player.show_info()

  elif select_menu == 3:
     if player is None:
        print("캐릭터를 생성후 사용해주세요~")
     elif player.hp <= 0:
        print("캐릭터가 사망하여 전투할 수 없습니다.")
     else:
        monster = Player("mob", 50, 5)
        turn_count = 0

        while monster.hp > 0 and player.hp > 0:           

          turn_count += 1    

          monster.take_damage(player.attack_power)

          if monster.hp > 0:
              player.take_damage(monster.attack_power)

          print(
              f"{turn_count}턴 "
              f"{monster.name} HP {monster.hp} / "
              f"{player.name} HP {player.hp}"
          )

        # 전투 결과
        if player.hp <= 0:
          print("전투 패배!")
        else:
          print("전투 승리!")

  elif select_menu == 4:
    if player is None:
      print("캐릭터를 생성후 사용해주세요~")
    else:
      print("회복량을 입력하세요:")

      try:
        healValue = int(input())       
      except ValueError:
        print(f"숫자만 입력해주세요~!")
      else:
        try:
          player.heal(healValue)                
        except ValueError as e:
          print(f"양수만 입력해주세요~! -> {e}")

  elif select_menu == 5:
      if player is None:
        print("플레이어 생성하고 저장해라")
      else:
        player.save("player")
      
  elif select_menu == 6:      
      load_player = Player("", 0, 0)
      try:
        load_player.load("player")
      except FileNotFoundError:
        print("저장된 파일이 없다.")        
      except (json.JSONDecodeError, KeyError, TypeError, ValueError) as e
        print(f"json 저장 데이터의 오류 : {e}")
      else:
        print("불러오기 성공!!")
        player = load_player
        player.show_info()

  else:
    print("잘못된 메뉴 번호입니다.")

print("프로그램이 종료되었습니다.")
```

```python
def get_grade(score):
  if type(score) is int:
    if score < 0 or score > 100:
      raise ValueError("0 미만 또는 100 초과 점수를 입력 할 수 없습니다.")
    else:
      if score >= 90:
        return "A"
      elif score >= 80:
        return "B"
      elif score >= 70:
        return "C"
      elif score >= 60:
        return "D"
      else:
        return "F"
  else:
    raise ValueError("정수만 입력 허용합니다.")

  
print(get_grade(95))
print(get_grade(82))
print(get_grade(73))
print(get_grade(65))
print(get_grade(40))

try:
  print(get_grade(120))
except ValueError as e:
  print(e)

try:
  print(get_grade("90"))
except ValueError as e:
  print(e)

```

```python
def analyze_scores(scores):   
  average =  sum(scores) / len(scores) 
  max_value = max(scores)
  min_value = min(scores)
  passed = 0
  excellent = []

  for i in scores:
    if i >= 60:
      passed += 1
    if i >= 90:
      excellent.append(i)

  result = {"average":average,
          "max":max_value,
          "min":min_value,
          "passed":passed,
          "excellent":excellent
          }
  
  return result   


scores = [95, 70, 82, 45, 100, 68, 77]
analyze_scores(scores)

```
