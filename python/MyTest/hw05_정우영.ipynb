* 문제 1 
  - 문자열을 입력받아 모음(a e i o u)의 개수를 출력하세요.  


 


```python
input_value = input("문자열을 입력하세요>").lower()
target_count = 0

for value in input_value:
  if value in "aeiou":
      target_count += 1

print(target_count)
```

* 문제 2 
  - 다음 딕셔너리가 있다.

    scores = {
        "Kim": 80,
        "Lee": 92,
        "Park": 68,
        "Choi": 100
    }

    다음을 출력하시오.

  - 평균 점수
  - 최고 점수 학생
  - 80점 이상 학생 이름

```python
scores = {
        "Kim": 80,
        "Lee": 92,
        "Park": 68,
        "Choi": 100
    }

average = sum(scores.values()) / len(scores)
print("평균 점수:", average)

sort_result = sorted(scores.items(), 
                     key=lambda item: item[1], 
                     reverse=True)

print(f"최고 점수 학생 : {sort_result[0][0]}")

for name, score in scores.items():
  if score >= 80:
    print(f"80점 이상 학생 이름 : {name}")
```

*  문제3) 
  다음 리스트에서

  numbers = [1,2,2,3,5,3,7,8,7,10]

  중복을 제거한 뒤 오름차순으로 출력하시오.



```python
numbers = [1,2,2,3,5,3,7,8,7,10]
sort_result = sorted(numbers, reverse=False)
numbers.clear()

for value in sort_result:
  if  value not in numbers:
    numbers.append(value)

print(numbers)
```

* 문제4) 문장을 입력받아 

I like python because python is fun

다음처럼 출력하시오.(딕셔너리 활용)

I : 1
like : 1
python : 2
because : 1
is : 1
fun : 1

```python
input_str = input("문장을 입력하세요>")

split_result = input_str.split(" ")
dict_result = {}

for value in split_result:
  if value in dict_result:
    dict_result[value] += 1
  else:
    dict_result[value] = 1

for key, value in dict_result.items():
  print(f"{key} : {value}")
```
