```python
class Student:
  def __init__(self, name, korean, math, english, science):
    self.name = name
    self.korean = korean
    self.math = math
    self.english = english
    self.science = science

  def get_sum(self):
    return self.korean + self.math + self.english + self.science

  def get_average(self):
    return self.get_sum() / 4

  def to_string(self):
    return f"{self.name}\t{self.get_sum()}\t{self.get_average()}"

a = Student("우영", 1, 2, 3, 4)
b = Student("윤서", 5, 6, 7, 8)
c = Student("윤정", 9, 10, 11, 12)

list_student = [a,b,c]

for student in list_student:
   print(student.to_string())
```

```python
# 1. Player 클래스 작성
class Player:
    def __init__(self, name, hp):
        self.name = name
        self.hp = hp        

# 2. Monster 클래스 작성
class Monster:
    def __init__(self, name, attack):
        self.name = name
        self.attack = attack

# 3. NPC 클래스 작성
class NPC:
    def __init__(self, name, job):
        self.name = name
        self.job = job 

# 4. 리스트에 인스턴스 저장
characters = [
    Player("기사", 100),
    Monster("슬라임", 15),
    NPC("상인", "아이템 판매"),
]

# 5. isinstance()로 구분해 출력
for char in characters:
    if isinstance(char, Player):
        print(f"{char.name} {char.hp}")
    elif isinstance(char, Monster):
        print(f"{char.name} {char.attack}")
    elif isinstance(char, NPC):
        print(f"{char.name} {char.job}")
```

```python
class Character:
    count = 0
    characters = []

    def __init__(self, name, hp):        
        # 1. 인스턴스 속성 초기화
        self.name = name
        self.hp = hp

        # 2. count 증가
        Character.count += 1

        # 3. characters에 현재 객체 추가
        Character.characters.append(self)        

    def __str__(self):
        return f"{self.name} {self.hp}"

    @classmethod
    def print_all(cls):
        # 4. 등록된 모든 캐릭터 출력
        for char in cls.characters:            
            print(char)

Character("기사", 100)
Character("마법사", 70)
Character("궁수", 80)

Character.print_all()
```

```python
import math

class Circle:
  def __init__(self, radius):
    self.__radius = radius

  def get_circumference(self):
    return 2 * math.pi * self.__radius

  def get_area(self):
    return math.pi * self.__radius ** 2  

  def get_radius(self):
    return self.__radius

  def set_radius(self, radius):
      if radius > 0:
          self.__radius = radius
      else:
          print("반지름은 양수여야 합니다.")
    


circle = Circle(10)
print(circle.get_radius())


circle.set_radius(-5)
print(circle.get_radius())

```

```python
class Circle:
  def __init__(self, radius):
    self.__radius = radius

  @property
  def radius(self):
    return self.__radius

  @radius.setter
  def radius(self, value):
    if value > 0:
      self.__radius = value
    else:
      print("반지름은 양수여야 합니다.")

c = Circle(10)
c.radius = 1000
print(c.radius)
```

```python
class Character:
    def __init__(self, name, hp):
        self.name = name
        self.hp = hp

    @property
    def hp(self):
        return self.__hp

    @hp.setter
    def hp(self, value):
        if value >= 0 and value <= 100:
          self.__hp = value
        else:
          print("hp는 0~100 사이여야 합니다.")        

    def __str__(self):
        return f"{self.name} - HP: {self.hp}"


player = Character("기사", 100)

print(player)

player.hp = 70
print(player)

player.hp = -10
print(player)

player.hp = 150
print(player)
```

```python
# 1. 사용자 정의 예외 클래스
class ErrorForAttack(Exception):
    pass


# 2. 부모 클래스
class Character:
    def __init__(self, name, hp):
        self.name = name
        self.hp = hp

    def attack(self):
        print(f"{self.name} 기본공격!")       


# 3. 전사 클래스
class Warrior(Character):
    def __init__(self, name, hp, attack_power):
        super().__init__(name, hp)

        if attack_power <= 0:
          raise ErrorForAttack("오류: 공격력은 0보다 커야 합니다.")
        else:
          self.attack_power = attack_power      
        

    def attack(self):
        print(f"{self.name} 검 공격! 공격력: {self.attack_power}")        


# 4. 마법사 클래스
class Mage(Character):
    def __init__(self, name, hp, mp):
        super().__init__(name, hp)
        self.mp = mp                 

    def attack(self):
        print(f"{self.name} 마법 공격! mp: {self.mp}")        


# 5. 캐릭터 생성 및 공격 테스트
char = Character("일반인", 50)
char.attack()

knight = Warrior("기사", 50, 30)
knight.attack()

mage = Mage("마법사", 50, 100)
mage.attack()


try:
  knight2 = Warrior("기사", 50, 0)
  knight2.attack()
except ErrorForAttack as e:
  print(e)  


# 6. 잘못된 공격력 예외 처리
```

```python
class InvalidHPError(Exception):
    pass


class Character:
    count = 0

    def __init__(self, name, hp):
        self.name = name
        self.hp = hp
        Character.count += 1

    @property
    def hp(self):
        return self.__hp

    @hp.setter
    def hp(self, value):
        if value > 0 and value <= 100 :
            self.__hp = value
        else:            
            raise InvalidHPError("오류: HP는 0~100 사이여야 합니다.")

    def __str__(self):
        return f"{self.name} - HP: {self.hp}"


class Warrior(Character):
    def __init__(self, name, hp, attack_power):
        super().__init__(name, hp)
        self.attack_power = attack_power

    def attack(self):
        print(f"{self.name} 검 공격! 공격력: {self.attack_power}")        

class Mage(Character):
    def __init__(self, name, hp, mp):
        super().__init__(name, hp)
        self.mp = mp                 

    def attack(self):
        print(f"{self.name} 마법 공격! mp: {self.mp}")   


warrior = Warrior("기사", 100, 30)
mage = Mage("마법사", 70, 100)

characters = [warrior, mage]

for char in characters:
    print(char)
    char.attack()

# 클래스 변수로 생성된 캐릭터 수 확인
print(f"총 캐릭터 수:{Character.count}")

# 잘못된 HP를 가진 캐릭터 생성 시 예외 처리
try:
    warrior2 = Warrior("기사2", -100, 30)
except InvalidHPError as e:
    print(e)
```
