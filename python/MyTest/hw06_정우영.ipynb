* 문제 1 
  - 좌표들이 저장된 튜플 리스트에서 x좌표의 합과 y좌표의 합을 구하세요.

      points = [(1,2), (3,4), (5,6)]

    출력: x의 합 : 9
          y의 합 : 12

```python
points = [(1,2), (3,4), (5,6)]

sum_x = 0
sum_y = 0

for point in points:
  sum_x += point[0]
  sum_y += point[1]

print(f"x 합 : {sum_x}")
print(f"y 합 : {sum_y}")
```

* 문제 2
  - numbers = [4, 10, 7, 3, 15, 8]
    정렬 함수(sort(), sorted())를 사용하지 않고 두 번째로 큰 수를 찾으세요.

```python
numbers = [4, 10, 7, 3, 15, 8]

first = 0
second = 0

for number in numbers:
    if number > first:
        second = first
        first = number
    elif second < number < first:
        second = number

print(f"두 번째로 큰 수 {second}")  # 10
```

* 문제 3
  - 비밀번호는 다음 조건을 만족해야 합니다.

  - 길이 8자 이상
  - 숫자가 하나 이상 포함
  - 영문자가 하나 이상 포함

    조건을 만족하면

    "사용 가능한 비밀번호입니다."

    그렇지 않으면

    "사용할 수 없는 비밀번호입니다."

    를 출력하세요.

```python
pass_word = input("비밀번호를 입력하세요>")

has_digit = False
has_letter = False
has_len_count = (len(pass_word) >= 8)

for char in pass_word:
    if "0" <= char <= "9":
        has_digit = True
    elif "a" <= char <= "z" or "A" <= char <= "Z":
        has_letter = True

if has_digit and has_letter and has_len_count:
    print("사용가능")
else:
    print("사용불가 ")
```

* 문제 4
 - 숫자를 입력받아 소수인지 판별하는 함수를 작성하세요.

    입력
    17

    출력
    소수입니다.
  

```python
# 소수는 1보다 크고 1과 자기 자신으로만 나누어 떨어지는 수라고 함!! 
def check_num(number):
  if number < 2:
    return False

  for value in range(2, number):
        if number % value == 0:
            return False

  return True  

is_check_num = check_num(int(input("숫자를 입력하세요>")))

if is_check_num == True:
  print("소수!")
else:
  print("소수가 아닙니다.")
```

* 문제5)
  - words = ["python", "java", "c", "javascript"]

  각 단어의 길이를 하나씩 반환하는 제너레이터를 작성하세요.

  출력
  6
  4
  1
  10


```python
words = ["python", "java", "c", "javascript"]

def word_lens(words):
    for word in words:
        yield len(word)

it = word_lens(words)
for i in range(len(words)):
  print(next(it))    

print()

for length in word_lens(words):
    print(length)
  
```
