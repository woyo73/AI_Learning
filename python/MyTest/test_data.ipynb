```python
import numpy as np

# 리스트와 numpy 배열의 차이
a = np.array([10, 20, 30, 40, 50])
b = [10, 20, 30, 40, 50]

print(a * 2)
print(b * 2)
```

```python
import numpy as np

a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print(a+b)
print(a-b)
print(a*b)
print(a/b)
```

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])

# 배열의 차원수 
print(a.ndim)

# 각 차원의 크기 
print(a.shape)

# 전체 요소 수 
print(a.size)

# 데이터 타입 
print(a.dtype)
```
