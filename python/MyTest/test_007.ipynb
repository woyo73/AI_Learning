```python


import numpy as np

damages = np.array([10, 20, 30, 40, 50])

print("합계 :", np.sum(damages))
print("평균 :", np.mean(damages))
print("최대 :", np.min(damages))
print("최소 :", np.max(damages))
```

```python
import numpy as np

stats = np.array([
  [100, 20], 
  [80, 30],
  [60, 50]
])

print(np.sum(stats, axis=0))
print(np.sum(stats, axis=1))
print(np.mean(stats, axis=0))
print(np.mean(stats, axis=1))
```

```python
import numpy as np

stats = np.array([
  [100, 20],
  [80, 30],
  [60, 50]
])

bonus = np.array([10, 5])
new_stats = stats + bonus
print(new_stats)

print(np.sum(new_stats, axis=0))
print(np.mean(new_stats, axis=0))

print(np.max(new_stats[:,1]))
print(np.min(new_stats[:,1]))

# new_stats = np.reshape(new_stats, (2, 3))
result = new_stats.reshape(2, 3)

print(result)


```
