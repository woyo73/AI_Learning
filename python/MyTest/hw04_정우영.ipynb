* 문제 3
  - numbers = [12, 7, 25, 9, 18, 30] 
  - 합계 / 평균 / 최댓값 / 최솟값 / 짝수 개수 / 홀수 개수 
  - sum(), max() 등을 사용하지 않고 직접 반복문으로 구현해보세요.

```python
numbers = [12, 7, 25, 9, 18, 30] 

sum = 0
odd_count = 0
even_count = 0

for i in numbers:
  if i % 2 == 0:
    even_count += 1
  else :
    odd_count += 1      
  sum += i

average = sum / len(numbers)
min = sorted(numbers)[0]
max = sorted(numbers,reverse=True)[0]

print(f"합계:{sum}");
print(f"평균:{average}");
print(f"최댓값:{max}");
print(f"최솟값:{min}");
print(f"홀수 개수:{odd_count}");
print(f"짝수 개수:{even_count}");
```

# 문제 4
  - 피라미드

```python
max_line_count = 25
star_count = 1
space_count = max_line_count - 1

for y in range(max_line_count):
    for x in range(space_count):
        print(" ", end="")

    for x in range(star_count):
        print("*", end="")

    star_count += 2
    space_count -= 1

    print()
```
