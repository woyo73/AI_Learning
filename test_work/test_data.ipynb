```python
import numpy as np

# 리스트와 numpy 배열의 차이
a = np.array([10, 20, 30, 40, 50])
b = [10, 20, 30, 40, 50]

print(a * 2)
print(b * 2)
```

```python
import numpy as np

a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print(a+b)
print(a-b)
print(a*b)
print(a/b)
```

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])

# 배열의 차원수 
print(a.ndim)

# 각 차원의 크기 
print(a.shape)

# 전체 요소 수 
print(a.size)

# 데이터 타입 
print(a.dtype)
```

```python
import numpy as np

scores = np.array([70, 80, 90, 100])
print(type(scores))

print(scores+10)

a = [1, 2, 3, 4]
b = [10, 20, 30, 40]
c = []

for index, num in enumerate(a):  
  c.append(num * b[index])

print(c)


numbers = np.array([[10, 20, 30],
                    [40, 50, 60]])
print(numbers.ndim)
print(numbers.shape)
print(numbers.size)


scores = np.array([60, 70, 80, 90, 100])

# 평균 
print(scores.mean())

# 최대값 
print(scores.max())

# 최소값 
print(scores.min())
```

```python
import numpy as np

attack = np.array([100, 80, 120, 90])
defense = np.array([20, 30, 40, 10])
damage = attack - defense

print(damage)
```

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])

print(a[0])    # 10
print(a[2])    # 30
print(a[-1])   # 50


a = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(a[0, 0])  # 10
print(a[1, 2])  # 60
print(a[2, 1])  # 80
```
