- 다음 분석을 수행해 보세요.

* 상품별 5일간 총판매 수량

* 상품별 총매출액

* 전체 상품의 총매출액

* 하루 평균 판매 수량

* 가장 많이 판매된 상품

* 매출액이 가장 높은 상품

```python
import numpy as np

# 한 쇼핑몰에서 3개 상품을 5일 동안 판매한 수량이 다음과 같다고 가정해 보자.

sales = np.array([
    [10, 15, 12, 20, 18],
    [5, 8, 10, 7, 12],
    [20, 18, 25, 22, 30]
])

prices = np.array([10000, 20000, 5000])


# 상품별 5일간 총 판매 수량 ()
print(f"상품별 5일간 총 판매 수량 : {sales.sum(axis=1)}")

# * 상품별 총매출액
print(f"상품별 총매출액 : {sales.sum(axis=1) * prices}")

# 전체 상품의 총매출액
print(f"전체 상품의 총매출액 : {np.sum((sales.sum(axis=1) * prices))}")

# 하루 평균 판매 수량
print(f"하루 평균 판매 수량 : {sales.mean(axis=0)}")

# 가장 많이 판매된 상품 인덱스
print(f"가장 많이 판매된 상품 행번호 : {np.argsort(sales.sum(axis=1))[::-1]}")

# 매출액이 가장 높은 상품
print(f"매출액이 가장 높은 상품 행번호 : {np.argsort(sales.sum(axis=1) * prices)[::-1]}")





```
