```python
import tensorflow as tf
import numpy as np
```

```python
# 수업시간, 자습시간 
x_data = np.array([
            [1, 2],
            [2, 3],
            [3, 1],
            [4, 3],
            [5, 3],
            [6, 2]
         ])
y_data = np.array([[0], [0],[0],[1],[1],[1]])
```

```python
model = tf.keras.Sequential()

# 입력 설정 
model.add(tf.keras.Input(shape=(2,)))

# 출력 설정
model.add(tf.keras.layers.Dense(
    units=1,
    activation='sigmoid'
))
```

```python
# 3. 학습 설정
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.05),
    # optimizer='sgd',
    loss='binary_crossentropy',                 
    metrics=['accuracy']
)

model.summary()
```

```python
# 4. 학습

loss = model.fit(
    x_data,
    y_data,
    epochs=500,
    verbose=1
)

```

```python
# # 5. 새로운 데이터 예측
x_test = np.array([[2, 2], [4, 0], [5, 1], [7, 5]])
predictions = model.predict(x_test, verbose=1)
```

```python
# 6. 확률을 0 또는 1로 분류
for x, p in zip(x_test, predictions):
    probability = p[0]
    result = 1 if probability >= 0.5 else 0

    print(f"공부시간:{x[0]:.0f}  자습시간:{x[1]:.0f}")
    print(f"합격 확률: {probability * 100:.2f}%")
    print(f"예측 결과: {'합격' if result == 1 else '불합격'}")
    print("----------------")
```

```python

print(f"Accuracy: {loss.history['accuracy'][-1] * 100:.2f}%")
```
