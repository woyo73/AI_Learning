# 신경망 구현 순서  

1. Sequential 모형 클래스 객체 생성.  
2. add 메소드로 레이어(layer) 추가.  
   - 입력단부터 순차적으로 추가한다.
   - 레이어는 출력(units) 뉴런 갯수를 첫번째 인수로 받는다.
   - 최초의 레이어는 Input(shape=( , ))로 입력 크기를 설정해야 한다.
   - activation 인수로 활성화 함수 설정한다.
3. compile 메소드로 모형 완성.
   - loss 인수로 비용(cost)함수 설정.
   - optimizer 인수로 최적화 알고리즘 설정
   - metrics 인수로 트레이닝 단계에서 기록할 성능 기준 설정
4. fit 메소드로 트레이닝
   - epochs로 에포크(epoch) 횟수 설정

```python
# 텐서플로우 설치
# !pip install tensorflow
import tensorflow as tf

print(tf.__version__)
```

```python
import numpy as np

x_train = np.array([1., 2., 3., 4.])
y_train = np.array([0., -1., -2., -3.])

# x_mean = np.mean(x_np)
# y_mean = np.mean(y_np)

# 기울기 
# w = np.sum((x_np - x_mean) * (y_np - y_mean)) / np.sum((x_np - x_mean) ** 2)

# 절편 x가 0일때 y값 
# b = y_mean - w * x_mean

# print("기울기 W:", w)
# print("절편 b:", b)

# new_x = 9
# predicted_y = w * new_x + b

# print(f"new_x{new_x} -> 예측값:{predicted_y}")
```

```python
# 1. Sequential 모형 클래스 객체 생성.
# 신경망 객체 생성
model = tf.keras.Sequential()
```

```python
# 2. add 메소드로 레이어(layer) 추가. 

# 입력 레이어 추가 
model.add(tf.keras.Input(shape=(1,)))

# 출력 레이어 추가 -> activation 는 linear(선형 회기)
model.add(tf.keras.layers.Dense(units=1, activation='linear'))
```

```python
# 3. compile 메소드로 모형 완성.

sgd = tf.keras.optimizers.SGD(learning_rate=0.1)

model.compile(
  optimizer=sgd,                  # 사용할 최적화 알고리즘 sgd:확률적 경사하강법
  loss='mse'                      # 손실함수 
)
```

```python
model.summary()
```

```python
# 학습 전, 최초 설정된 w 값 조회
start_weights = model.layers[0].get_weights()
print(f"학습전 : w {start_weights[0]}, b {start_weights[1]}")
start_weights
```

```python
# 4. fit 메소드로 트레이닝
# 학습 
model.fit(x_train, y_train, epochs=200)
```

```python
# 학습된 모델에 새로운 X 입력
# x_test = np.array([-100., 22., 24.])
x_test = np.array([10., 11., 12.])

# Y 예측
y_predict = model.predict(x_test)
y_predict
# prediction
# print("예측값:", y_predict[0][0])
```

```python
# 첫 번 째 레이어에 존재하는 w 와 b

# 학습을 통해 구한 w와 b의 값
end_weights = model.layers[0].get_weights()
print(f"학습후 : w {end_weights[0]}, b {end_weights[1]}")
print(model.layers[0].weights)
```
